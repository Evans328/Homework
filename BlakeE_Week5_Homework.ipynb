15. Using `Series`, create a `DataFrame` that looks like this:

    | Ingredients | Quantity | Unit |
    |----|----|----|
    | Flour | 4 | cups |
    | Milk | 1 | cup |
    | Eggs | 2 | large |
    | Spam | 1 | can |

    Name: Dinner, dtype: object

````python
import pandas as pd

# Columns 
ingredients = pd.Series(["Flour", "Milk", "Eggs", "Spam"])
quantity = pd.Series([4, 1, 2, 1])
unit = pd.Series(["cups", "cup", "large", "can"])

# Combine
df = pd.DataFrame({
    "Ingredients": ingredients,
    "Quantity": quantity,
    "Unit": unit
})

# Display
print(df)
````

### Or ###

````python
import pandas as pd

data = {
    "Ingredients": pd.Series(["Flour", "Milk", "Eggs", "Spam"], index=[0, 1, 2, 3]),
    "Quantity": pd.Series([4, 1, 2, 1], index=[0, 1, 2, 3]),
    "Unit": pd.Series(["cups", "cup", "large", "can"], index=[0, 1, 2, 3])
}

dinner_df = pd.DataFrame(data)
print(dinner_df)
````

        ["ice cream", "Cashew Fudge", "waffels", "Carrot Halwa"]
18. Sort the 'studentData' DataFrame in Ascending order -- Sorting by column 'Name' and then "address"

````python
import pandas as pd

# 16. Create the DataFrame named studentData
data = {
    'Name': ['Jai', 'janusha', 'Gaurav', 'Anuj'],
    'Height': [5.1, 6.2, 5.1, 5.2],
    'Qualification': ['Msc', 'MA', 'Msc', 'Msc'],
    'address': ['Delhi', 'Doha', 'Chennai', 'Dakhar'],
    'Age': [21, 23, 24, 21],
    'Pets': ['Dog', 'Bunny', 'Chinchilla', 'Parrot'],
    'sport': ['Darts', 'Basketball', 'PaddleBoarding', 'Cricket']
}
studentData = pd.DataFrame(data)

# 17. Add a new column named 'Desserts'
studentData['Desserts'] = ["ice cream", "Cashew Fudge", "waffels", "Carrot Halwa"]

# 18. Sort by 'Name' and then 'address' in Ascending order
# 'janusha' will sort after uppercase names because python sorting is case-sensetive
studentData.sort_values(by=['Name', 'address'], ascending=True, inplace=True)

# Display the final sorted DataFrame
print(studentData)
````
