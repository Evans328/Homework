```python
from collections import Counter


raw_tags = [
    "  Bug ",
    "bug",
    "Feature",
    " feature",
    "UI",
    "Bug",
    "UI  ",
    "Feature",
]


clean_tags = [tag.strip().capitalize() for tag in raw_tags]


counts = Counter(clean_tags)
print("--- QUICK INSIGHTS DASHBOARD ---")
for category, count in counts.items():
  bar = "█" * count
  print(f"{category:<10} | {bar} ({count})")
```

```python
from collections import Counter
import re


data_stream = "The tabby cat chased the mouse, but the other tabby cat fell asleep."


cleaned_data = re.sub(r'[^\w\s]', '', data_stream.lower())
words = cleaned_data.split()


word_counts = Counter(words)

print(f"📊 DATA INSIGHTS (Total Words: {len(words)})")
print("-" * 35)
for word, count in word_counts.most_common(3):
    print(f"🔹 Word: '{word:<6}' | Count: {count:<2} | Frequency: {count/len(words):.1%}")
```
